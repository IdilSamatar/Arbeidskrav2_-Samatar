# Oppgave 5


```python
import numpy as np

def beregn_figur(a, b):

    areal_halvsirkel = np.pi * (a/2)**2 / 2
    omkrets_halvsirkel = (2 * np.pi * (a/2)) / 2

    areal_trekant = (a * b) / 2
    skrå_side_trekant = np.sqrt(a**2 + b**2)

    ytre_omkrets_figur = omkrets_halvsirkel + a + skrå_side_trekant

    areal_figur = areal_halvsirkel + areal_trekant

    return (areal_figur, ytre_omkrets_figur)


a = float(input("Skriv inn lengden på a: "))
b = float(input("Skriv inn lengden på b: "))

areal, omkrets = beregn_figur(a, b)

print("Arealet av figuren er:", areal)
print("Den ytre omkretsen av figuren er:", omkrets)
```
